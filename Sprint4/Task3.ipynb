<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Task в C#, который будет представлять задачи внутри
проекта. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма

Требования к базовому классу Task:

• Атрибуты: ID задачи (TaskId), Название задачи (TaskName), Приоритет
задачи (Priority).

Методы:

o MarkAsComplete(): метод для отметки задачи как выполненной.

o GetTaskDetails(): метод для получения деталей задачи.

o ReassignTo(): метод для переназначения задачи другому члену
команды.

Требования к производным классам:
1. ДелегатскаяЗадача (DelegateTask): Должна содержать дополнительные
атрибуты, такие как Дата выполнения (DueDate).
Метод MarkAsComplete() должен быть переопределен для включения даты
выполнения в сообщение о завершении задачи.
2. КоманднаяЗадача (TeamTask): Должна содержать дополнительные атрибуты,
такие как Команда (TeamName). Метод ReassignTo() должен быть
переопределен для указания нового члена команды, которому будет
переназначена задача.
3. ИсследовательскаяЗадача (ResearchTask) (если требуется третий класс):
Должна содержать дополнительные атрибуты, такие как Исходные данные
(DataSource). Метод GetTaskDetails() должен быть переопределен для
отображения источников данных, используемых в задаче, вместе с другими
деталями задачи.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями

```
using System;
using System.Collections.Generic;

// ==== Интерфейсы ====
public interface ITrackable
{
    void ShowStatus();
}

public interface IPrioritizable
{
    void UpgradePriority(string newPriority);
}

public interface ITaskService
{
    void AddTask(Task task);
    void DisplayAllTasks();
    Task FindById(int id);
}

// ==== Базовый класс Task ====
public class Task : ITrackable, IPrioritizable
{
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    public string AssignedTo { get; private set; }
    public bool IsCompleted { get; private set; }
    public Task RelatedTask { get; private set; }

    // Новые атрибуты
    public DateTime CreatedDate { get; set; }
    public string Description { get; set; }
    public int EstimatedHours { get; set; }
    public string Category { get; set; }
    public string Location { get; set; }
    public bool RequiresApproval { get; set; }

    // Дополнительные атрибуты
    public decimal Budget { get; set; }
    public List<string> Tags { get; set; }
    public int Complexity { get; set; }
    public DateTime Deadline { get; set; }

    public Task(int taskId, string taskName, string priority, string assignedTo = "Не назначено")
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        AssignedTo = assignedTo;
        IsCompleted = false;
        CreatedDate = DateTime.Now;
        Description = "Описание отсутствует";
        EstimatedHours = 0;
        Category = "Общее";
        Location = "Офис";
        RequiresApproval = false;
        
        // Инициализация новых атрибутов
        Budget = 0;
        Tags = new List<string>();
        Complexity = 1;
        Deadline = DateTime.Now.AddDays(7);
    }

    // Новые методы
    public virtual void UpdateBudget(decimal amount)
    {
        Budget = amount;
        Console.WriteLine($"Бюджет задачи '{TaskName}' обновлен: {amount} руб.");
    }

    public virtual void AddTag(string tag)
    {
        Tags.Add(tag);
        Console.WriteLine($"Добавлен тег '{tag}' к задаче '{TaskName}'");
    }

    public virtual void SetComplexity(int level)
    {
        Complexity = level;
        Console.WriteLine($"Сложность задачи '{TaskName}' установлена: {level}");
    }

    public virtual void ExtendDeadline(int days)
    {
        Deadline = Deadline.AddDays(days);
        Console.WriteLine($"Срок задачи '{TaskName}' продлен на {days} дней. Новый срок: {Deadline.ToShortDateString()}");
    }

    // Существующие методы
    public virtual void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"Задача '{TaskName}' выполнена.");
    }

    public virtual void GetTaskDetails()
    {
        Console.WriteLine($"| {TaskId,-3} | {TaskName,-20} | {Priority,-8} | {(IsCompleted ? "Да" : "Нет"),-10} | {AssignedTo,-12} | {(RelatedTask != null ? RelatedTask.TaskName : "—"),-20} | {EstimatedHours,3}ч | {CreatedDate.ToShortDateString(),-10} | {Description,-20} | {Category,-10} | {Location,-10} | {RequiresApproval,-5} | {Budget,8} руб. | {Complexity,2} | {Deadline.ToShortDateString(),-10} |");
    }

    public virtual void ReassignTo(string newMember)
    {
        AssignedTo = newMember;
        Console.WriteLine($"Задача '{TaskName}' переназначена на {AssignedTo}.");
    }

    public void ReassignTo(string newMember, string newPriority)
    {
        AssignedTo = newMember;
        Priority = newPriority;
        Console.WriteLine($"Задача '{TaskName}' переназначена на {AssignedTo} с приоритетом {Priority}.");
    }

    public void SetRelatedTask(Task task)
    {
        RelatedTask = task;
    }

    // ==== Явная реализация интерфейсов ====
    void ITrackable.ShowStatus()
    {
        Console.WriteLine($"[ITrackable] Статус задачи '{TaskName}': {(IsCompleted ? "Выполнена" : "В процессе")}");
    }

    void IPrioritizable.UpgradePriority(string newPriority)
    {
        Priority = newPriority;
        Console.WriteLine($"[IPrioritizable] Приоритет задачи '{TaskName}' обновлён на {Priority}");
    }

    // Обычные методы для вызова напрямую
    public void ShowStatus()
    {
        Console.WriteLine($"Статус задачи '{TaskName}': {(IsCompleted ? "Выполнена" : "В процессе")}");
    }

    public void UpgradePriority(string newPriority)
    {
        Priority = newPriority;
        Console.WriteLine($"Приоритет задачи '{TaskName}' обновлён на {Priority}");
    }
}

// ==== Производные классы ====
public class DelegateTask : Task
{
    public DateTime DueDate { get; set; }
    public string DelegateTo { get; set; }
    public string Reminder { get; set; }
    public bool IsUrgent { get; set; }
    public string Notes { get; set; }

    // Новые атрибуты
    public string DelegationReason { get; set; }
    public int DelegationLevel { get; set; }
    public bool RequiresConfirmation { get; set; }

    public DelegateTask(int taskId, string taskName, string priority, DateTime dueDate, string assignedTo = "Не назначено")
        : base(taskId, taskName, priority, assignedTo)
    {
        DueDate = dueDate;
        DelegateTo = "Не назначено";
        Reminder = "Нет";
        IsUrgent = false;
        Notes = "";
        
        // Инициализация новых атрибутов
        DelegationReason = "Не указана";
        DelegationLevel = 1;
        RequiresConfirmation = true;
    }

    // Новые методы
    public void SetDelegationReason(string reason)
    {
        DelegationReason = reason;
        Console.WriteLine($"Причина делегирования задачи '{TaskName}': {reason}");
    }

    public void SetDelegationLevel(int level)
    {
        DelegationLevel = level;
        Console.WriteLine($"Уровень делегирования задачи '{TaskName}' установлен: {level}");
    }

    public void RequestConfirmation()
    {
        if (RequiresConfirmation)
        {
            Console.WriteLine($"Запрос подтверждения делегирования задачи '{TaskName}' отправлен");
        }
    }

    public override void MarkAsComplete()
    {
        base.MarkAsComplete();
        Console.WriteLine($"Дата выполнения: {DateTime.Now.ToShortDateString()}, срок: {DueDate.ToShortDateString()}");
    }

    public void Delegate(string member)
    {
        DelegateTo = member;
        Console.WriteLine($"Задача '{TaskName}' делегирована на {DelegateTo}");
    }

    public void Delegate(string member, DateTime newDueDate)
    {
        DelegateTo = member;
        DueDate = newDueDate;
        Console.WriteLine($"Задача '{TaskName}' делегирована на {DelegateTo} со сроком {DueDate.ToShortDateString()}");
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"   Доп. информация: Причина делегирования - {DelegationReason}, Уровень - {DelegationLevel}");
    }
}

public class TeamTask : Task
{
    public string TeamName { get; set; }
    public int TeamSize { get; set; }
    public string Sprint { get; set; }
    public bool IsCritical { get; set; }
    public string ToolsRequired { get; set; }

    // Новые атрибуты
    public List<string> TeamMembers { get; set; }
    public string ProjectManager { get; set; }
    public int CompletionPercentage { get; set; }

    public TeamTask(int taskId, string taskName, string priority, string teamName, int teamSize, string assignedTo = "Не назначено")
        : base(taskId, taskName, priority, assignedTo)
    {
        TeamName = teamName;
        TeamSize = teamSize;
        Sprint = "Не определено";
        IsCritical = false;
        ToolsRequired = "Стандартные";
        
        // Инициализация новых атрибутов
        TeamMembers = new List<string>();
        ProjectManager = "Не назначен";
        CompletionPercentage = 0;
    }

    // Новые методы
    public void AddTeamMember(string member)
    {
        TeamMembers.Add(member);
        TeamSize++;
        Console.WriteLine($"В команду {TeamName} добавлен {member}. Новый размер: {TeamSize}");
    }

    public void SetProjectManager(string manager)
    {
        ProjectManager = manager;
        Console.WriteLine($"Руководитель проекта для задачи '{TaskName}' назначен: {manager}");
    }

    public void UpdateCompletion(int percentage)
    {
        CompletionPercentage = percentage;
        Console.WriteLine($"Прогресс выполнения задачи '{TaskName}': {percentage}%");
    }

    public override void ReassignTo(string newMember)
    {
        base.ReassignTo(newMember);
        Console.WriteLine($"(Командная задача) Теперь {newMember} из команды {TeamName} выполняет '{TaskName}'. Размер команды: {TeamSize}");
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"   Команда: {TeamName}, Руководитель: {ProjectManager}, Прогресс: {CompletionPercentage}%");
    }
}

public class ResearchTask : Task, ITrackable, IPrioritizable
{
    public string DataSource { get; set; }
    public string ResearchMethod { get; set; }
    public bool RequiresInternalApproval { get; set; }
    public string Field { get; set; }
    public int SampleSize { get; set; }

    // Новые атрибуты
    public string Hypothesis { get; set; }
    public List<string> References { get; set; }
    public bool IsPeerReviewed { get; set; }

    public ResearchTask(int taskId, string taskName, string priority, string dataSource, string researchMethod, string assignedTo = "Не назначено")
        : base(taskId, taskName, priority, assignedTo)
    {
        DataSource = dataSource;
        ResearchMethod = researchMethod;
        RequiresInternalApproval = true;
        Field = "Область исследования";
        SampleSize = 0;
        
        // Инициализация новых атрибутов
        Hypothesis = "Не сформулирована";
        References = new List<string>();
        IsPeerReviewed = false;
    }

    // Новые методы
    public void SetHypothesis(string hypothesis)
    {
        Hypothesis = hypothesis;
        Console.WriteLine($"Гипотеза исследования '{TaskName}' установлена: {hypothesis}");
    }

    public void AddReference(string reference)
    {
        References.Add(reference);
        Console.WriteLine($"Добавлена ссылка в исследование '{TaskName}': {reference}");
    }

    public void MarkAsPeerReviewed()
    {
        IsPeerReviewed = true;
        Console.WriteLine($"Исследование '{TaskName}' отмечено как рецензированное");
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"Источник данных: {DataSource}, Метод: {ResearchMethod}, Поле: {Field}, Размер выборки: {SampleSize}");
        Console.WriteLine($"   Гипотеза: {Hypothesis}, Рецензировано: {(IsPeerReviewed ? "Да" : "Нет")}, Ссылок: {References.Count}");
    }

    // Явная реализация интерфейсов для ResearchTask
    void ITrackable.ShowStatus()
    {
        Console.WriteLine($"[ITrackable-Research] Статус исследовательской задачи '{TaskName}': {(IsCompleted ? "Завершено" : "В процессе исследования")}");
    }

    void IPrioritizable.UpgradePriority(string newPriority)
    {
        Priority = newPriority;
        Console.WriteLine($"[IPrioritizable-Research] Приоритет исследования '{TaskName}' повышен до {Priority}");
    }

    // Обычные методы для ResearchTask
    public new void ShowStatus()
    {
        Console.WriteLine($"Статус исследования '{TaskName}': {(IsCompleted ? "Завершено" : "В процессе исследования")}");
    }

    public new void UpgradePriority(string newPriority)
    {
        Priority = newPriority;
        Console.WriteLine($"Приоритет исследования '{TaskName}' повышен до {Priority}");
    }
}

// ==== Generic-класс коллекции ====
public class TaskCollection<T> where T : Task
{
    private readonly List<T> _tasks = new List<T>();

    public void Add(T task)
    {
        _tasks.Add(task);
    }

    public void Remove(T task)
    {
        _tasks.Remove(task);
    }

    public T Find(Predicate<T> predicate)
    {
        return _tasks.Find(predicate);
    }

    public void DisplayTasks()
    {
        Console.WriteLine(new string('-', 280));
        Console.WriteLine($"| {"ID",-3} | {"Название",-20} | {"Приоритет",-8} | {"Выполнена",-10} | {"Назначена",-12} | {"Связана с",-20} | {"Часы",-5} | {"Создано",-10} | {"Описание",-20} | {"Категория",-10} | {"Локация",-10} | {"Одобрено",-5} | {"Бюджет",-8} | {"Слож.",-2} | {"Срок",-10} |");
        Console.WriteLine(new string('-', 280));
        foreach (var task in _tasks)
        {
            task.GetTaskDetails();
        }
        Console.WriteLine(new string('-', 280));
    }
}

// ==== Сервис для DI ====
public class TaskService : ITaskService
{
    private readonly TaskCollection<Task> _collection;

    public TaskService(TaskCollection<Task> collection)
    {
        _collection = collection;
    }

    public void AddTask(Task task)
    {
        _collection.Add(task);
        Console.WriteLine($"Задача '{task.TaskName}' добавлена в сервис");
    }

    public void DisplayAllTasks()
    {
        Console.WriteLine("\n=== ВСЕ ЗАДАЧИ ЧЕРЕЗ СЕРВИС ===");
        _collection.DisplayTasks();
    }

    public Task FindById(int id)
    {
        var task = _collection.Find(t => t.TaskId == id);
        if (task != null)
        {
            Console.WriteLine($"Найдена задача с ID {id}: {task.TaskName}");
        }
        else
        {
            Console.WriteLine($"Задача с ID {id} не найдена");
        }
        return task;
    }
}

// ==== Создание и использование задач ====

// Создание задач
Task task1 = new Task(1, "Обычная задача", "Низкий", "Иван");
DelegateTask task2 = new DelegateTask(2, "Сделать отчёт", "Высокий", DateTime.Now.AddDays(3), "Анна");
TeamTask task3 = new TeamTask(3, "Разработка модуля", "Средний", "Команда А", 5, "Мария");
ResearchTask task4 = new ResearchTask(4, "Анализ рынка", "Низкий", "Open Data Portal", "Статистический метод", "Алексей");

// Использование новых методов
Console.WriteLine("=== ИСПОЛЬЗОВАНИЕ НОВЫХ МЕТОДОВ ===");

// Для базовой задачи
task1.UpdateBudget(15000);
task1.AddTag("Срочно");
task1.AddTag("Важно");
task1.SetComplexity(3);
task1.ExtendDeadline(5);

// Для делегируемой задачи
task2.SetDelegationReason("Перераспределение нагрузки");
task2.SetDelegationLevel(2);
task2.RequestConfirmation();

// Для командной задачи
task3.AddTeamMember("Ольга");
task3.AddTeamMember("Петр");
task3.SetProjectManager("Сергей");
task3.UpdateCompletion(45);

// Для исследовательской задачи
task4.SetHypothesis("Рынок вырастет на 20% в следующем квартале");
task4.AddReference("Market Analysis 2024");
task4.AddReference("Industry Report Q3");
task4.MarkAsPeerReviewed();

// устанавливаем связи между задачами
task2.SetRelatedTask(task1);
task3.SetRelatedTask(task2);
task4.SetRelatedTask(task3);

// создаём коллекцию
TaskCollection<Task> allTasks = new TaskCollection<Task>();

// добавляем задачи
allTasks.Add(task1);
allTasks.Add(task2);
allTasks.Add(task3);
allTasks.Add(task4);

// выводим таблицу задач
allTasks.DisplayTasks();

// базовые операции из вашего кода
Console.WriteLine("\n=== БАЗОВЫЕ ОПЕРАЦИИ ===");
task1.ShowStatus();
task2.Delegate("Сергей", DateTime.Now.AddDays(5));
task3.AddTeamMember("Ольга");
task4.UpgradePriority("Высокий");

// ==== Новые пункты ====
Console.WriteLine("\n=== НОВЫЕ ОПЕРАЦИИ ===");

// 1️⃣ Назначим описание, часы и категорию
task1.Description = "Проверка системы логов";
task1.EstimatedHours = 5;
task1.Category = "Тестирование";
task1.Location = "Москва";
task1.RequiresApproval = true;

// 2️⃣ Обновим задачу и покажем детали заново
task1.GetTaskDetails();

// 3️⃣ Переназначим задачу с новым приоритетом
task1.ReassignTo("Павел", "Средний");

// 4️⃣ Отметим выполненную задачу
task1.MarkAsComplete();
task3.MarkAsComplete();

// 5️⃣ Используем сервис для DI и поиска задачи по ID
TaskCollection<Task> diCollection = new TaskCollection<Task>();
TaskService service = new TaskService(diCollection);

service.AddTask(task1);
service.AddTask(task2);
service.AddTask(task3);
service.AddTask(task4);

Console.WriteLine("\n=== ПОИСК ЗАДАЧИ С ID = 3 ЧЕРЕЗ СЕРВИС ===");
Task found = service.FindById(3);
if (found != null)
{
    found.GetTaskDetails();
}

// 6️⃣ Показать весь список задач через сервис
Console.WriteLine("\n=== ОТОБРАЖЕНИЕ ВСЕХ ЗАДАЧ ЧЕРЕЗ TASKSERVICE ===");
service.DisplayAllTasks();

// 7️⃣ Явное использование интерфейсов (принудительно)
Console.WriteLine("\n=== ИНТЕРФЕЙСНЫЕ ВЫЗОВЫ ===");
((ITrackable)task3).ShowStatus();
((IPrioritizable)task4).UpgradePriority("Критический");

// 8️⃣ Демонстрация переопределения методов в наследниках
Console.WriteLine("\n=== ПРОВЕРКА ПЕРЕОПРЕДЕЛЁННЫХ МЕТОДОВ ===");
task2.MarkAsComplete();
task3.ReassignTo("Виктор");

// 9️⃣ Дополнительная демонстрация новых возможностей
Console.WriteLine("\n=== ДОПОЛНИТЕЛЬНЫЕ ВОЗМОЖНОСТИ ===");
task4.UpdateBudget(50000);
task2.AddTag("Делегировано");
task3.ExtendDeadline(10);
```
