```polyglot-notebook

public class Product
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    public int Quantity { get; set; }
    public string Category { get; set; }
    public string Description { get; set; }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity = Quantity + amount;
    }

    public string GetProductInfo()
    {
        return Name + " (" + Category + "): " + Quantity + " шт., " + Price + " руб. — " + Description;
    }
}

public class Seller
{
    public string Name { get; set; }
    public string EmployeeId { get; set; }
    protected decimal Salary { get; set; }
    internal string ContactInfo { get; set; }

    public Product[] Products = new Product[5];
    public int ProductCount = 0;

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        Products[ProductCount] = product;
        ProductCount++;
    }

    public void SellProduct(Product product, int quantity)
    {
        product.UpdateQuantity(-quantity);
        Console.WriteLine(Name + " продал " + quantity + " шт. " + product.Name);
    }

    public string GetSellerInfo()
    {
        return "Продавец: " + Name + ", ID: " + EmployeeId + ", Контакт: " + ContactInfo + ", Зарплата: " + Salary + " руб.";
    }
}

public class Store
{
    private string StoreName;
    private string Location;
    public string StoreHours { get; set; }
    public Seller[] Sellers = new Seller[5];
    public int SellerCount = 0;

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        Sellers[SellerCount] = seller;
        SellerCount++;
    }

    public void ListProducts()
    {
        for (int i = 0; i < SellerCount; i++)
        {
            Seller s = Sellers[i];
            for (int j = 0; j < s.ProductCount; j++)
            {
                Console.WriteLine(s.Products[j].GetProductInfo());
            }
        }
    }

    public string GetStoreInfo()
    {
        return "Магазин: " + StoreName + ", Адрес: " + Location + ", Часы работы: " + StoreHours;
    }
}

Store store = new Store("Магазин", "ул. Ленина", "9:00 - 21:00");
Seller seller = new Seller("Александра", "S001", 50000, "sashatalalaeva4@gmail.com");
Product product = new Product("Телефон", 25000, 5, "Электроника", "Смартфон Samsung");

seller.AddProduct(product);
store.AddSeller(seller);

Console.WriteLine(store.GetStoreInfo());
Console.WriteLine(seller.GetSellerInfo());
store.ListProducts();
```
