 Онлайн-приложение для посещения музеев в виртуальном формате.



```polyglot-notebook
using System;
using System.Collections.Generic; 

// Описывает музейный экспонат виртуального музея
public class Exhibit
{
    // Уникальный идентификатор экспоната
    public int Id { get; set; }

    // Название экспоната
    public string Title { get; set; }

    // Описание экспоната
    public string Description { get; set; }

    // Историческая эпоха
    public string Era { get; set; }

    // Автор или создатель экспоната
    public string Author { get; set; }

    // Метод для вывода информации об экспонате
    public void ShowInfo()
    {
        Console.WriteLine($"Экспонат: {Title}");
        Console.WriteLine($"Автор: {Author}");
        Console.WriteLine($"Эпоха: {Era}");
        Console.WriteLine($"Описание: {Description}");
        Console.WriteLine();
    }
}

// Описывает виртуальный экскурсионный тур
public class Tour
{
    // Идентификатор тура
    public int Id { get; set; }

    // Название тура
    public string Name { get; set; }

    // Продолжительность тура (в минутах)
    public int Duration { get; set; }

    // Список экспонатов, входящих в тур
    public List<Exhibit> Exhibits { get; set; } = new List<Exhibit>();

    // Добавление экспоната в тур
    public void AddExhibit(Exhibit exhibit)
    {
        Exhibits.Add(exhibit);
    }

    // Удаление экспоната из тура
    public void RemoveExhibit(Exhibit exhibit)
    {
        Exhibits.Remove(exhibit);
    }

    // Отображение информации о туре и его экспонатах
    public void ShowTour()
    {
        Console.WriteLine($"Тур: {Name}");
        Console.WriteLine($"Продолжительность: {Duration} минут");
        Console.WriteLine("Экспонаты тура:");

        foreach (var exhibit in Exhibits)
        {
            exhibit.ShowInfo();
        }
    }
}

// Описывает пользователя виртуального музея
public class Visitor
{
    // Идентификатор посетителя
    public int Id { get; set; }

    // Имя посетителя
    public string Name { get; set; }

    // Электронная почта посетителя
    public string Email { get; set; }

    // Тур, выбранный посетителем
    public Tour SelectedTour { get; set; }

    // Выбор тура
    public void ChooseTour(Tour tour)
    {
        SelectedTour = tour;
        Console.WriteLine($"{Name} выбрал(а) тур \"{tour.Name}\"");
    }

    // Начало виртуального тура
    public void StartTour()
    {
        // Проверка: выбран ли тур
        if (SelectedTour == null)
        {
            Console.WriteLine("Тур не выбран.");
            return;
        }

        Console.WriteLine($"Посетитель {Name} начинает тур:");
        SelectedTour.ShowTour();
    }
}

// Используется для создания и хранения виртуальных туров
public class TourManager
{
    // Список всех доступных туров
    public List<Tour> Tours { get; set; } = new List<Tour>();

    // Создание нового тура
    public void CreateTour(Tour tour)
    {
        Tours.Add(tour);
        Console.WriteLine($"Тур \"{tour.Name}\" создан.");
    }

    // Отображение всех доступных туров
    public void ShowAllTours()
    {
        Console.WriteLine("Доступные туры:");
        foreach (var tour in Tours)
        {
            Console.WriteLine($"- {tour.Name} ({tour.Duration} мин.)");
        }
    }
}

// Создание первого экспоната
Exhibit exhibit1 = new Exhibit
{
    Id = 1,
    Title = "Мона Лиза",
    Author = "Леонардо да Винчи",
    Era = "Эпоха Возрождения",
    Description = "Одна из самых известных картин в мире."
};

// Создание второго экспоната
Exhibit exhibit2 = new Exhibit
{
    Id = 2,
    Title = "Давид",
    Author = "Микеланджело",
    Era = "Эпоха Возрождения",
    Description = "Мраморная скульптура эпохи Возрождения."
};

// Создание виртуального тура
Tour renaissanceTour = new Tour
{
    Id = 1,
    Name = "Шедевры Возрождения",
    Duration = 45
};

// Добавление экспонатов в тур
renaissanceTour.AddExhibit(exhibit1);
renaissanceTour.AddExhibit(exhibit2);

// Создание менеджера туров
TourManager tourManager = new TourManager();

// Добавление тура в систему
tourManager.CreateTour(renaissanceTour);

// Вывод списка всех доступных туров
tourManager.ShowAllTours();

// Создание посетителя виртуального музея
Visitor visitor = new Visitor
{
    Id = 1,
    Name = "Анна",
    Email = "anna@mail.com"
};

// Посетитель выбирает тур
visitor.ChooseTour(renaissanceTour);

// Посетитель начинает виртуальный тур
visitor.StartTour();
```
