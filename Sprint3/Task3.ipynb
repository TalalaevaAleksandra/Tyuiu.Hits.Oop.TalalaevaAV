### Вариант задания 


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Task в C#, который будет представлять задачи внутри
проекта. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма

Требования к базовому классу Task:

• Атрибуты: ID задачи (TaskId), Название задачи (TaskName), Приоритет
задачи (Priority).

Методы:

o MarkAsComplete(): метод для отметки задачи как выполненной.

o GetTaskDetails(): метод для получения деталей задачи.

o ReassignTo(): метод для переназначения задачи другому члену
команды.

Требования к производным классам:
1. ДелегатскаяЗадача (DelegateTask): Должна содержать дополнительные
атрибуты, такие как Дата выполнения (DueDate).
Метод MarkAsComplete() должен быть переопределен для включения даты
выполнения в сообщение о завершении задачи.
2. КоманднаяЗадача (TeamTask): Должна содержать дополнительные атрибуты,
такие как Команда (TeamName). Метод ReassignTo() должен быть
переопределен для указания нового члена команды, которому будет
переназначена задача.
3. ИсследовательскаяЗадача (ResearchTask) (если требуется третий класс):
Должна содержать дополнительные атрибуты, такие как Исходные данные
(DataSource). Метод GetTaskDetails() должен быть переопределен для
отображения источников данных, используемых в задаче, вместе с другими
деталями задачи.


#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

// ==== Интерфейсы ====
public interface ITrackable
{
    void ShowStatus();
}

public interface IPrioritizable
{
    void UpgradePriority(string newPriority);
}

// ==== Базовый класс Task ====
public class Task : ITrackable, IPrioritizable
{
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    public string AssignedTo { get; private set; }
    public bool IsCompleted { get; private set; }
    public Task RelatedTask { get; private set; }

    // Новые атрибуты
    public DateTime CreatedDate { get; set; }
    public string Description { get; set; }
    public int EstimatedHours { get; set; }
    public string Category { get; set; }
    public string Location { get; set; }
    public bool RequiresApproval { get; set; }

    public Task(int taskId, string taskName, string priority, string assignedTo = "Не назначено")
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        AssignedTo = assignedTo;
        IsCompleted = false;
        CreatedDate = DateTime.Now;
        Description = "Описание отсутствует";
        EstimatedHours = 0;
        Category = "Общее";
        Location = "Офис";
        RequiresApproval = false;
    }

    public virtual void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"Задача '{TaskName}' выполнена.");
    }

    public virtual void GetTaskDetails()
    {
        Console.WriteLine($"| {TaskId,-3} | {TaskName,-20} | {Priority,-8} | {(IsCompleted ? "Да" : "Нет"),-10} | {AssignedTo,-12} | {(RelatedTask != null ? RelatedTask.TaskName : "—"),-20} | {EstimatedHours,3}ч | {CreatedDate.ToShortDateString(),-10} | {Description,-20} | {Category,-10} | {Location,-10} | {RequiresApproval,-5} |");
    }

    public virtual void ReassignTo(string newMember)
    {
        AssignedTo = newMember;
        Console.WriteLine($"Задача '{TaskName}' переназначена на {AssignedTo}.");
    }

    public void ReassignTo(string newMember, string newPriority)
    {
        AssignedTo = newMember;
        Priority = newPriority;
        Console.WriteLine($"Задача '{TaskName}' переназначена на {AssignedTo} с приоритетом {Priority}.");
    }

    public void SetRelatedTask(Task task) => RelatedTask = task;
    public void ShowStatus() => Console.WriteLine($"Статус задачи '{TaskName}': {(IsCompleted ? "Выполнена" : "В процессе")}");
    public void UpgradePriority(string newPriority) { Priority = newPriority; Console.WriteLine($"Приоритет задачи '{TaskName}' обновлён на {Priority}"); }
}

// ==== Производные классы ====
public class DelegateTask : Task
{
    public DateTime DueDate { get; set; }
    public string DelegateTo { get; set; }
    public string Reminder { get; set; }
    public bool IsUrgent { get; set; }
    public string Notes { get; set; }

    public DelegateTask(int taskId, string taskName, string priority, DateTime dueDate, string assignedTo = "Не назначено")
        : base(taskId, taskName, priority, assignedTo)
    {
        DueDate = dueDate;
        DelegateTo = "Не назначено";
        Reminder = "Нет";
        IsUrgent = false;
        Notes = "";
    }

    public override void MarkAsComplete()
    {
        base.MarkAsComplete();
        Console.WriteLine($"Дата выполнения: {DateTime.Now.ToShortDateString()}, срок: {DueDate.ToShortDateString()}");
    }

    public void Delegate(string member) { DelegateTo = member; Console.WriteLine($"Задача '{TaskName}' делегирована на {DelegateTo}"); }
    public void Delegate(string member, DateTime newDueDate) { DelegateTo = member; DueDate = newDueDate; Console.WriteLine($"Задача '{TaskName}' делегирована на {DelegateTo} со сроком {DueDate.ToShortDateString()}"); }
}

public class TeamTask : Task
{
    public string TeamName { get; set; }
    public int TeamSize { get; set; }
    public string Sprint { get; set; }
    public bool IsCritical { get; set; }
    public string ToolsRequired { get; set; }

    public TeamTask(int taskId, string taskName, string priority, string teamName, int teamSize, string assignedTo = "Не назначено")
        : base(taskId, taskName, priority, assignedTo)
    {
        TeamName = teamName;
        TeamSize = teamSize;
        Sprint = "Не определено";
        IsCritical = false;
        ToolsRequired = "Стандартные";
    }

    public override void ReassignTo(string newMember)
    {
        base.ReassignTo(newMember);
        Console.WriteLine($"(Командная задача) Теперь {newMember} из команды {TeamName} выполняет '{TaskName}'. Размер команды: {TeamSize}");
    }

    public void AddTeamMember(string member)
    {
        TeamSize++;
        Console.WriteLine($"В команду {TeamName} добавлен {member}. Новый размер: {TeamSize}");
    }
}

public class ResearchTask : Task, ITrackable, IPrioritizable
{
    public string DataSource { get; set; }
    public string ResearchMethod { get; set; }
    public bool RequiresApproval { get; set; }
    public string Field { get; set; }
    public int SampleSize { get; set; }

    public ResearchTask(int taskId, string taskName, string priority, string dataSource, string researchMethod, string assignedTo = "Не назначено")
        : base(taskId, taskName, priority, assignedTo)
    {
        DataSource = dataSource;
        ResearchMethod = researchMethod;
        RequiresApproval = true;
        Field = "Область исследования";
        SampleSize = 0;
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"Источник данных: {DataSource}, Метод исследования: {ResearchMethod}, Поле: {Field}, Размер выборки: {SampleSize}");
    }
}

// ==== Generic-класс коллекции ====
public class TaskCollection<T> where T : Task
{
    private List<T> _tasks = new List<T>();
    public void Add(T task) => _tasks.Add(task);
    public void Remove(T task) => _tasks.Remove(task);

    public void DisplayTasks()
    {
        Console.WriteLine(new string('-', 220));
        Console.WriteLine($"| {"ID",-3} | {"Название",-20} | {"Приоритет",-8} | {"Выполнена",-10} | {"Назначена",-12} | {"Связана с",-20} | {"Часы",-5} | {"Создано",-10} | {"Описание",-20} | {"Категория",-10} | {"Локация",-10} | {"Одобрено",-5} |");
        Console.WriteLine(new string('-', 220));
        foreach (var task in _tasks)
            task.GetTaskDetails();
        Console.WriteLine(new string('-', 220));
    }
}

Task task1 = new Task(1, "Обычная задача", "Низкий", "Иван");
DelegateTask task2 = new DelegateTask(2, "Сделать отчёт", "Высокий", DateTime.Now.AddDays(3), "Анна");
TeamTask task3 = new TeamTask(3, "Разработка модуля", "Средний", "Команда А", 5, "Мария");
ResearchTask task4 = new ResearchTask(4, "Анализ рынка", "Низкий", "Open Data Portal", "Статистический метод", "Алексей");

task2.SetRelatedTask(task1);
task3.SetRelatedTask(task2);
task4.SetRelatedTask(task3);

TaskCollection<Task> allTasks = new TaskCollection<Task>();
allTasks.Add(task1);
allTasks.Add(task2);
allTasks.Add(task3);
allTasks.Add(task4);

allTasks.DisplayTasks();

task1.ShowStatus();
task2.Delegate("Сергей", DateTime.Now.AddDays(5));
task3.AddTeamMember("Ольга");
task4.UpgradePriority("Высокий");
```
