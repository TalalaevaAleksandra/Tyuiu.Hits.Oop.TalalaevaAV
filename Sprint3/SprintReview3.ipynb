```polyglot-notebook
using System;
using System.Collections.Generic;
using System.Linq;
// Базовый абстрактный класс Component
public abstract class Component
{
    public string Manufacturer { get; set; }
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string manufacturer, string name, decimal price)
    {
        Manufacturer = manufacturer;
        Name = name;
        Price = price;
    }
// Виртуальный метод GetInfo
    public virtual string GetInfo()
    {
        return ($"{Name} от {Manufacturer} по цене {Price} руб.");
    }
// Перегрузка оператора ToString для удобного представления
    public override string ToString()
    {
        return GetInfo();
    }
}
// Релизация интерфейса IInstallable
public interface IInstallable
{
    void Install();
}
// Релизация класса, наследующийся от Component
public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double Frequency { get; set; }
    public CPU(string manufacturer, string name, decimal price, int cores, double frequency)
        : base(manufacturer, name, price)
    {
        Cores = cores;
        Frequency = frequency;
    }
// Переопределение метода GetInfo
    public override string GetInfo()
    {
        return $"{base.GetInfo()}, ядер: {Cores}";
    }
// Реализация интерфейса IInstallable
    public void Install()
    {
        Console.WriteLine($"Установка процессора: {Name} в сокет...");
    }
    // Перегрузка операторов сравнения (требование №2.2)
    public static bool operator ==(CPU left, CPU right)
    {
        if (ReferenceEquals(left, right)) return true;
        if (left is null || right is null) return false;
        return left.Name == right.Name && left.Manufacturer == right.Manufacturer && left.Cores == right.Cores && left.Frequency == right.Frequency;
    }

    public static bool operator !=(CPU left, CPU right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU cpu && this == cpu;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Cores, Frequency);
    }
}

public class RAM : Component, IInstallable
{
    public int Capacity { get; set; }
    public string Type { get; set; }

    public RAM(string name, string manufacturer, decimal price, int capacity, string type) 
        : base(name, manufacturer, price)
    {
        Capacity = capacity;
        Type = type;
    }

    // Перекрытие метода GetInfo()
    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Объем: {Capacity}ГБ, Тип: {Type}";
    }

    // Реализация интерфейса IUnstallable
    public void Install()
    {
        Console.WriteLine($"Установка оперативной памяти: {Name} в слот...");
    }

    // Перегрузка операторов сравнения
    public static bool operator ==(RAM left, RAM right)
    {
        if (ReferenceEquals(left, right)) return true;
        if (left is null || right is null) return false;
        return left.Name == right.Name && left.Manufacturer == right.Manufacturer && left.Capacity == right.Capacity && left.Type == right.Type;
    }

    public static bool operator !=(RAM left, RAM right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM ram && this == ram;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Capacity, Type);
    }
}

public class GPU : Component, IInstallable
{
    public int Memory { get; set; }
    public string MemoryType { get; set; }

    public GPU(string name, string manufacturer, decimal price, int memory, string memoryType) 
        : base(name, manufacturer, price)
    {
        Memory = memory;
        MemoryType = memoryType;
    }

    // Перекрытие метода GetInfo()
    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Память: {Memory}ГБ {MemoryType}";
    }

    // Реализация интерфейса IUnstallable
    public void Install()
    {
        Console.WriteLine($"Установка видеокарты: {Name} в слот PCIe...");
    }

    // Перегрузка операторов сравнения
    public static bool operator ==(GPU left, GPU right)
    {
        if (ReferenceEquals(left, right)) return true;
        if (left is null || right is null) return false;
        return left.Name == right.Name && left.Manufacturer == right.Manufacturer && left.Memory == right.Memory && left.MemoryType == right.MemoryType;
    }

    public static bool operator !=(GPU left, GPU right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU gpu && this == gpu;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Memory, MemoryType);
    }
}

public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }
    public int Speed { get; set; }

    public HDD(string name, string manufacturer, decimal price, int capacity, int speed) 
        : base(name, manufacturer, price)
    {
        Capacity = capacity;
        Speed = speed;
    }

    // Перекрытие метода GetInfo()
    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Объем: {Capacity}ГБ, Скорость: {Speed}об/мин";
    }

    // Реализация интерфейса IUnstallable
    public void Install()
    {
        Console.WriteLine($"Установка жесткого диска: {Name} в отсек...");
    }

    // Перегрузка операторов сравнения
    public static bool operator ==(HDD left, HDD right)
    {
        if (ReferenceEquals(left, right)) return true;
        if (left is null || right is null) return false;
        return left.Name == right.Name && left.Manufacturer == right.Manufacturer && left.Capacity == right.Capacity && left.Speed == right.Speed;
    }

    public static bool operator !=(HDD left, HDD right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD hdd && this == hdd;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Capacity, Speed);
    }
}

// 4. Класс Computer (основной класс системы)
public class Computer
{
    private List<Component> components = new List<Component>();

    // Generic метод для добавления компонента (требование №3.1)
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент {component.Name} успешно добавлен!\n");
    }

    // Generic метод для удаления компонента (требование №3.2)
    public bool RemoveComponent<T>(T component) where T : Component
    {
        var result = components.Remove(component);
        if (result)
        {
            Console.WriteLine($"Компонент {component.Name} успешно удален!\n");
        }
        else
        {
            Console.WriteLine($"Компонент {component.Name} не найден!\n");
        }
        return result;
    }

    // Метод для получения общей стоимости (требование №3.3)
    public decimal GetTotalPrice()
    {
        return components.Sum(component => component.Price);
    }

    // Метод для отображения информации о всех компонентах
    public void DisplayComponents()
    {
        Console.WriteLine("Компоненты компьютера:");
        foreach (var component in components)
        {
            Console.WriteLine($"  - {component.GetInfo()}");
        }
        Console.WriteLine($"Общая стоимость: {GetTotalPrice()} руб.\n");
    }
}

// 5. Статический класс ComputerFactory (дополнительное задание №1)
public static class ComputerFactory
{
    public static Computer CreateGamingComputer()
    {
        var computer = new Computer();
        
        var cpu = new CPU("Ryzen 7 5800X", "AMD", 35000m, 8, 3.8);
        var ram = new RAM("Vengeance LPX", "Corsair", 12000m, 16, "DDR4");

var gpu = new GPU("RTX 4070", "NVIDIA", 60000m, 12, "GDDR6X");
        var hdd = new HDD("Barracuda", "Seagate", 8000m, 2000, 7200);

        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(gpu);
        computer.AddComponent(hdd);

        return computer;
    }

    public static Computer CreateOfficeComputer()
    {
        var computer = new Computer();
        
        var cpu = new CPU("Core i5-12400", "Intel", 20000m, 6, 2.5);
        var ram = new RAM("ValueRAM", "Kingston", 6000m, 8, "DDR4");
        var hdd = new HDD("Blue", "WD", 5000m, 1000, 5400);

        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(hdd);

        return computer;
    }

    public static Computer CreateBudgetComputer()
    {
        var computer = new Computer();
        
        var cpu = new CPU("Athlon 3000G", "AMD", 8000m, 2, 3.5);
        var ram = new RAM("Essential", "Crucial", 4000m, 4, "DDR4");
        var hdd = new HDD("Green", "WD", 4500m, 500, 5400);

        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(hdd);

        return computer;
    }
}

        // Демонстрация создания компьютеров через фабрику
        Console.WriteLine(" ComputerFactory:\n");

        var gamingPc = ComputerFactory.CreateGamingComputer();
        Console.WriteLine("Игровой компьютер:");
        gamingPc.DisplayComponents();

        var officePc = ComputerFactory.CreateOfficeComputer();
        Console.WriteLine("Офисный компьютер:");
        officePc.DisplayComponents();

        var budgetPc = ComputerFactory.CreateBudgetComputer();
        Console.WriteLine("Бюджетный компьютер:");
        budgetPc.DisplayComponents();

        // Демонстрация работы с компонентами
        Console.WriteLine(" Демонстрация операций с компонентами:\n");

        var customPc = new Computer();
        
        // Добавление компонентов
        var customCpu = new CPU("Core i7-13700K", "Intel", 42000m, 16, 3.4);
        var customRam = new RAM("Trident Z", "G.Skill", 15000m, 32, "DDR5");
        var customGpu = new GPU("RX 7800 XT", "AMD", 55000m, 16, "GDDR6");

        customPc.AddComponent(customCpu);
        customPc.AddComponent(customRam);
        customPc.AddComponent(customGpu);

        Console.WriteLine("Пользовательский компьютер до удаления:");
        customPc.DisplayComponents();

        // Удаление компонента
        customPc.RemoveComponent(customGpu);

        Console.WriteLine("Пользовательский компьютер после удаления видеокарты:");
        customPc.DisplayComponents();

        // Демонстрация сравнения компонентов
        Console.WriteLine("3. Демонстрация сравнения компонентов:\n");

        var cpu1 = new CPU("Ryzen 5 5600X", "AMD", 25000m, 6, 3.7);
        var cpu2 = new CPU("Ryzen 5 5600X", "AMD", 25000m, 6, 3.7);
        var cpu3 = new CPU("Core i5-12600K", "Intel", 28000m, 10, 3.7);

        Console.WriteLine($"CPU1: {cpu1.GetInfo()}");
        Console.WriteLine($"CPU2: {cpu2.GetInfo()}");
        Console.WriteLine($"CPU3: {cpu3.GetInfo()}");

        Console.WriteLine($"\nCPU1 == CPU2: {cpu1 == cpu2}");
        Console.WriteLine($"CPU1 == CPU3: {cpu1 == cpu3}");
        Console.WriteLine($"CPU1 != CPU3: {cpu1 != cpu3}");

        // Демонстрация полиморфизма
        Console.WriteLine("\n4. Демонстрация полиморфизма:\n");

        List<Component> components = new List<Component>
        {
            new CPU("Ryzen 9 7950X", "AMD", 70000m, 16, 4.5),
            new RAM("Dominator Platinum", "Corsair", 20000m, 64, "DDR5"),
            new GPU("RTX 4090", "NVIDIA", 160000m, 24, "GDDR6X"),
            new HDD("IronWolf Pro", "Seagate", 12000m, 4000, 7200)
        };

        Console.WriteLine("Полиморфный список компонентов:");

foreach (var component in components)
        {
            // Демонстрация полиморфного вызова GetInfo()
            Console.WriteLine($"  - {component.GetInfo()}");
        }

        


```
